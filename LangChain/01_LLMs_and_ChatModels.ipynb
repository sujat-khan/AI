## 1️⃣ ChatGroq — Free, Fast Llama & Mixtral Models

```python
from langchain_groq import ChatGroq

# ============================================================
# Available FREE Groq Models:
# - llama-3.1-8b-instant    → Fast, great for most tasks
# - llama-3.3-70b-versatile → Slower, higher quality
# - mixtral-8x7b-32768      → Good for long context (32K)
# - gemma2-9b-it            → Google's Gemma model
# ============================================================
llm = ChatGroq(
    model="llama-3.1-8b-instant",  # FREE model
    temperature=0.7,
    max_tokens=512,
)

print(f"Model: {llm.model_name}")
print(f"Temperature: {llm.temperature}")
```

```python
# Method 1: invoke() with a string
response = llm.invoke("What is the capital of India? Answer in one sentence.")

print("Type:", type(response).__name__)
print("Content:", response.content)
print("Model:", response.response_metadata.get('model', 'N/A'))
```

```python
# Method 2: invoke() with messages list
from langchain_core.messages import HumanMessage, SystemMessage

messages = [
    SystemMessage(content="You are a pirate. Always respond like a pirate."),
    HumanMessage(content="Tell me about Python programming.")
]

response = llm.invoke(messages)
print(response.content)
```

```python
# Method 3: stream() — Token-by-token streaming
print("Streaming response:\n")
print("-" * 50)

for chunk in llm.stream("Write a haiku about artificial intelligence."):
    print(chunk.content, end="", flush=True)

print("\n" + "-" * 50)
```

```python
# Method 4: batch() — Process multiple inputs
questions = [
    "What is Python in one sentence?",
    "What is JavaScript in one sentence?",
    "What is Rust in one sentence?"
]

responses = llm.batch(questions)

for q, r in zip(questions, responses):
    print(f"Q: {q}")
    print(f"A: {r.content}\n")
```

## 2️⃣ Different Groq Models — Quality vs Speed

```python
import time

models = {
    "llama-3.1-8b-instant":    "Fast, good for simple tasks",
    "llama-3.3-70b-versatile": "Slower, better quality",
    "mixtral-8x7b-32768":      "Good balance, 32K context",
    "gemma2-9b-it":            "Google's Gemma, compact",
}

prompt = "Explain recursion in programming in exactly 2 sentences."

print("Model Comparison (all FREE!)")
print("=" * 60)

for model, description in models.items():
    try:
        test_llm = ChatGroq(model=model, temperature=0)
        start = time.time()
        r = test_llm.invoke(prompt)
        elapsed = time.time() - start
        
        print(f"\n📦 {model} ({description})")
        print(f"   ⏱️  {elapsed:.2f}s")
        print(f"   💬 {r.content[:120]}...")
    except Exception as e:
        print(f"\n📦 {model}: Error - {str(e)[:50]}")
```

## 5️⃣ Structured Outputs — Works with ALL Free Providers

```python
from pydantic import BaseModel, Field
from typing import List

# Define output structure
class MovieReview(BaseModel):
    title: str = Field(description="Movie title")
    rating: float = Field(description="Rating from 1-10")
    pros: List[str] = Field(description="List of 3 positives")
    cons: List[str] = Field(description="List of 2 negatives")
    summary: str = Field(description="One sentence summary")

# Use Groq with structured output
llm = ChatGroq(model="llama-3.3-70b-versatile", temperature=0)
structured_llm = llm.with_structured_output(MovieReview)

review = structured_llm.invoke("Review the movie Inception (2010)")

print(f"Title:   {review.title}")
print(f"Rating:  {review.rating}/10")
print(f"Pros:    {review.pros}")
print(f"Cons:    {review.cons}")
print(f"Summary: {review.summary}")
```

```python
# ============================================================
# Async — works with all providers
# ============================================================
import asyncio

llm = ChatGroq(model="llama-3.1-8b-instant")

async def get_answers_concurrently(questions):
    tasks = [llm.ainvoke(q) for q in questions]
    responses = await asyncio.gather(*tasks)
    return [r.content for r in responses]

questions = ["What is Python?", "What is Go?", "What is Rust?"]
answers = await get_answers_concurrently(questions)

for q, a in zip(questions, answers):
    print(f"Q: {q}")
    print(f"A: {a[:80]}...\n")
```
