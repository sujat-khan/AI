## 📚 What is LangChain?

**LangChain** is an open-source framework designed to simplify the creation of applications using **Large Language Models (LLMs)**. It provides a standard interface to work with different LLMs, connect them to data sources, and build sophisticated AI pipelines (called **chains** and **agents**).

### 🌟 Why LangChain?

Before LangChain, building LLM-powered apps meant:
- Writing boilerplate code for every API call
- Manually managing conversation history
- Stitching together tools and data sources yourself
- No standard way to compose complex multi-step workflows

LangChain **solves all of this** with:

| Problem | LangChain Solution |
|---------|--------------------|
| Multiple LLM providers | Unified LLM interface |
| Prompt management | Prompt Templates |
| Multi-step workflows | Chains (LCEL) |
| Conversation history | Memory modules |
| External tools & APIs | Agents + Tools |
| Custom data ingestion | Document Loaders + Vector Stores |
| Monitoring & debugging | LangSmith integration |

| `langchain-anthropic` | Anthropic Claude integration |

## 🧩 Core Concepts — Quick Preview

### 1. The LCEL Pipeline Pattern

LangChain uses the **`|` pipe operator** to compose components, just like Unix pipes:

```python
chain = prompt | llm | output_parser
result = chain.invoke({"topic": "space"})
```

### 2. Runnable Interface

Every LangChain component implements the **Runnable** interface with these methods:

| Method | Description |
|--------|-------------|
| `.invoke(input)` | Run synchronously, return single output |
| `.stream(input)` | Stream output token by token |
| `.batch(inputs)` | Process multiple inputs in parallel |
| `.ainvoke(input)` | Async version of invoke |
| `.astream(input)` | Async streaming |
| `.abatch(inputs)` | Async batch processing |

### 3. The Schema System

LangChain uses **Pydantic** models for data validation:
- `HumanMessage`, `AIMessage`, `SystemMessage` — for chat
- `Document` — for loaded text data
- Custom schemas via `BaseModel`

````python
# Quick demo: LangChain message types
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage

# Creating messages
system_msg = SystemMessage(content="You are a helpful assistant.")
human_msg = HumanMessage(content="What is LangChain?")
ai_msg = AIMessage(content="LangChain is a framework for building LLM-powered apps!")

messages = [system_msg, human_msg, ai_msg]

for msg in messages:
    print(f"Type: {type(msg).__name__:15s} | Content: {msg.content}")
````

````python
# Quick demo: Document structure
from langchain_core.documents import Document

doc = Document(
    page_content="LangChain makes it easy to build LLM applications.",
    metadata={
        "source": "langchain_docs",
        "page": 1,
        "author": "Harrison Chase"
    }
)

print("Document Content:", doc.page_content)
print("Document Metadata:", doc.metadata)
````

## 🔑 Key Design Principles of LangChain

### 1. **Composability** — Build complex systems from simple parts
```python
chain = prompt | llm | parser  # Simple pipe composition
```

### 2. **Modularity** — Swap components without changing the rest
```python
# Switch from OpenAI to Anthropic — everything else stays the same!
chain_openai   = prompt | ChatGroq()   | parser
chain_anthropic = prompt | ChatAnthropic() | parser
```

### 3. **Observability** — Track, debug and evaluate
```python
# Every chain run is automatically logged to LangSmith
os.environ["LANGCHAIN_TRACING_V2"] = "true"
```

### 4. **Streaming-first** — Stream responses for better UX
```python
for chunk in chain.stream({"topic": "AI"}):
    print(chunk, end="", flush=True)
```
