# 🆓 LangChain for FREE — Setup Guide

---

## You do NOT need OpenAI to use LangChain!

Here are the **best free alternatives**:

---

## Option 1: 🚀 Groq (Recommended — Free & Very Fast)

**Groq** provides free API access to open-source models (Llama 3, Mixtral) with **blazing speed**.

### Steps:
1. Go to **[console.groq.com](https://console.groq.com)** → Sign up (free)
2. Go to **API Keys** → Create new API key
3. Copy the key → add to your `.env` file

**Free limits:** ~14,400 requests/day (very generous!)

---

## Option 2: 🔵 Google Gemini (Free Tier)

**Google AI Studio** gives you free Gemini access.

### Steps:
1. Go to **[aistudio.google.com](https://aistudio.google.com)** → Sign in with Google
2. Click **Get API Key** → Create API key
3. Copy the key → add to your `.env` file

**Free limits:** 15 RPM, 1 million tokens/day for Gemini 1.5 Flash

---

## Option 3: 🦙 Ollama (100% Local — No Internet, No Key!)

**Ollama** runs LLMs directly on your computer — completely free, private.

### Steps:
1. Download from **[ollama.com/download](https://ollama.com/download)**
2. Install and run it
3. Open terminal and run: `ollama pull llama3.2` (downloads ~2GB model)
4. No API key needed!

**Requirements:** 8GB+ RAM recommended

---

## Option 4: 🤗 HuggingFace (For Embeddings — Free & Local)

For **embeddings** (used in RAG), we use free local HuggingFace models.
**No API key, no cost, runs on your machine!**

## 🦙 Quick Test — Ollama (Local)

````python
# ============================================================
# Test Ollama — Runs 100% locally, zero cost
# Requires: ollama.com/download + `ollama pull llama3.2`
# ============================================================
from langchain_ollama import ChatOllama

ollama_llm = ChatOllama(
    model="llama3.2",  # Or any model you pulled
    temperature=0.7,
)

try:
    response = ollama_llm.invoke("Say hello in one sentence.")
    print("Ollama Response:")
    print(response.content)
except Exception as e:
    print(f"Ollama not running: {e}")
    print("Start Ollama: download from ollama.com, then run: ollama pull llama3.2")
````

## 🤗 Quick Test — HuggingFace Embeddings (Free & Local)

````python
# ============================================================
# Free local embeddings — No API key, no cost!
# Downloads model once (~90MB), then runs locally forever
# ============================================================
from langchain_huggingface import HuggingFaceEmbeddings

print("Loading embedding model (downloads ~90MB on first run)...")

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    model_kwargs={"device": "cpu"}  # Use "cuda" if you have a GPU!
)

# Test it
vector = embeddings.embed_query("Hello, LangChain!")
print(f"✅ Embeddings working!")
print(f"   Vector dimensions: {len(vector)}")
print(f"   First 5 values: {[round(v, 4) for v in vector[:5]]}")
````
